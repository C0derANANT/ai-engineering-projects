# 🏢 AI Company Brochure Generator
Generates a company brochure from its website using an LLM (Groq / Gemini), with a live preview and styled **PDF** and **DOCX** exports.

All the logic lives in the `brochure/` package; this notebook just lets you use it from Jupyter.

## Option 1: Use it in code

```python
from IPython.display import Markdown, display
from brochure import BrochureRequest, export_docx, export_pdf, fetch_website_context, stream_brochure

company = "Hugging Face"
website = fetch_website_context("https://huggingface.co")

request = BrochureRequest(company_name=company, tone="Professional", format="Mixed", length="Short", website=website)
for brochure_md in stream_brochure(request):
    pass  # streams the text; we only need the final version here

display(Markdown(brochure_md))
print("PDF :", export_pdf(brochure_md, company))
print("DOCX:", export_docx(brochure_md, company))
```
